**Introducción:**

[Chipotle Mexican Grill](https://www.chipotle.com) es una franquicia de restaurantes de comida rápida con sede en Estados Unidos. La especialidad de esta franquicia es comida Tex-Mex, es decir, gastronomía de la gente mexicano-estadounidense que vive en Texas. Uno de los productos estrella es el Burrito, el cual puede incluir una variedad de ingredientes.

En 2015, el New York Times publicó un artículo denominado "At Chipotle, How Many Calories Do People Really Eat?" ([enlace](https://www.nytimes.com/interactive/2015/02/17/upshot/what-do-people-actually-order-at-chipotle.html)), en donde se realiza un análisis sobre cuántas calorías, miligramos de sodio y gramos de grasa saturada están incluidos en las órdenes de dos restaurantes de Chipotle. El análisis presentado en el citado artículo tiene por base la información de cerca de 3000 comidas de alrededor de 1800 órdenes de [Grubhub](https://www.grubhub.com) de julio a diciembre de 2012 procedentes de dos restaurantes de Chipotle.
