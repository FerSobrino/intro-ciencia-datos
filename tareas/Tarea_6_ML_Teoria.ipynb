Vamos a utilizar el árbol de decisión que construimos para los datos de titanic. No edites el
siguiente código.

Responde las siguientes preguntas:

* ¿Qué significa para un modelo tener una alta accuracy pero baja precision o recall en este
  contexto?
* ¿Por qué podrían ser más informativas la precision y la recall que la exactitud para este
  conjunto de datos?
* ¿Cómo se relaciona el F1 score con la precision y la recall?

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import confusion_matrix

# Leemos los datos de Titanic
titanic_data = pd.read_csv("../data/otros/titanic.csv")
titanic_data.head()

# Crear conjunto de datos
features = ["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked"]
pred = "Survived"

# Separar covariables y variable a predecir
X = titanic_data[features]
y = titanic_data[pred]

# Imputaremos media para numéricas y moda para categóricas -
numerical_features = ["Age", "SibSp", "Parch", "Fare"]
categorical_features = ["Pclass", "Sex", "Embarked"]

X_num = X[numerical_features]
X_cat = X[categorical_features]

imputer_num = SimpleImputer(strategy='mean')
X_num_imputed = imputer_num.fit_transform(X_num)

imputer_cat = SimpleImputer(strategy='most_frequent')
X_cat_imputed = imputer_cat.fit_transform(X_cat)

# Transformaremos categóricas a dummies
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
X_cat_encoded = encoder.fit_transform(X_cat_imputed)
X_cat_encoded_df = pd.DataFrame(X_cat_encoded, columns=encoder.get_feature_names_out(categorical_features))

# Volvemos a juntar todo en un dataframe
X_processed = pd.concat([pd.DataFrame(X_num_imputed, columns=numerical_features), X_cat_encoded_df], axis=1)

# ahora sí vamos a separar en train y test
X_train, X_test, y_train, y_test = train_test_split(X_processed, y, test_size=0.2, random_state=42)

# Modelo Árbol de Decisión
arbol_clasificacion = DecisionTreeClassifier(max_depth=3, random_state=42)
arbol_clasificacion.fit(X_train, y_train)
y_pred = arbol_clasificacion.predict(X_test)
````

````python
# Matriz de confusion
cm = confusion_matrix(y_test, y_pred)

# Plot confusion matrix
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Not Survived', 'Survived'], yticklabels=['Not Survived', 'Survived'])
plt.ylabel('Real')
plt.xlabel('Pred')
plt.title('Confusion Matrix') # le voy a dejar el título en inglés porque los datos y las etiquetas están en inglés
plt.show()

# Métricas
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

# Print
print("Métricas:")
print(f'Accuracy: {accuracy:.2f}')
print(f'Precision: {precision:.2f}')
print(f'Recall: {recall:.2f}')
print(f'F1 Score: {f1:.2f}')
````
