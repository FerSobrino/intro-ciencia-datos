# Primera visualización

Usaremos los siguientes datos:

```python
url = "https://github.com/byuidatascience/data4python4ds/raw/master/data-raw/mpg/mpg.csv"
mpg = pd.read_csv(url)
mpg
```

```python
### Relación entre millas por galón y tamaño del motor
```

```python
mpg.plot.scatter(x='hwy', y='displ')
plt.title("Relación entre millas por galón y tamaño del motor")
plt.ylabel('millas por galón')
plt.xlabel('tamaño del motor')
plt.show()
```

### ¿Qué pasó acá?

- `.plot()` base python, por defecto genera una gráfica de línea de las x y las y's que le des.
- `pyplot` interface de `matplotlib` para mostrar las gráficas.
- `plt.show()` muestra la gráfica.

## ¿Qué tipos de gráficas puede hacer `.plot()`?

- `area`
- `bar`
- `barh`
- `box`
- `hexbin`
- `hist`
- `kde`
- `density`
- `line` (por defecto)
- `pie`
- `scatter`

- `.plot()` es un wrapper de `pyplot.plot()`.
- Puedes pasar el tipo de plot como parámetro o como método.

### Otras maneras de generar este plot:

```python
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(14, 6))

mpg.plot.scatter(x = 'hwy', y = 'displ', ax=axes[0])
mpg.plot(x = 'hwy', y = 'displ', kind = 'scatter', ax=axes[1])

plt.show()
```

### Agregar títulos y ejes

```python
mpg.plot.scatter(x = 'hwy', y = 'displ')
plt.title('''Relación entre millas por galón y 
tamaño del motor''')
plt.ylabel('millas por galón')
plt.xlabel('tamaño del motor')
plt.show()
```

## Cambiar aesthetic (color, tamaño, transparencia, etc )

```python
mpg.plot.scatter(x = 'hwy', y = 'displ', c = 'r', alpha = 0.5, 
marker = '>')
plt.title('''Relación entre millas por galón
y tamaño del motor''')
plt.ylabel('millas por galón')
plt.xlabel('tamaño del motor')
plt.show()
```

# Matplotlib y Seaborn

## ¿Qué es `matplotlib`?
* librería que permite crear figuras tanto estáticas como animadas
* usada principalmente para hacer gráficas rápidas durante la exploración de los datos 
* genera una figura con un conjunto de ejes sobre los que podemos poner nuestras gráficas/imágenes/etc
* dos niveles:
    * nivel figura: puedes modificar todo dentro de tu gráfica
    * nivel ejes: como construir la figura (acomodarlas, etc)
    

## Jerarquía de los objetos en `matplotlib`

* _Figure: contenedor de la gráfica, se puede definir antes o no definirlo y que se cree por default 
    * es como un lienzo vació
    * puede contener muchos objetos de tipo `axes`
    * dividir en subplots 
* _Axes_: una sola gráfica o imagen 
* Propiedades de una sola gráfica: leyendas, títulos, colores, markers, texto, etc

## Seaborn 

* Librería de visualización de datos 
* Está construida sobre `maptplotlib`
* Muchas gráficas se pueden hacer con `matplotlib` o `seaborn`: usa la más sencilla o la que se adapte mejor a los que necesites

# Otros tipos de gráficas

## ¿Le falta información a esta gráfica?

```python
sns.scatterplot(data = mpg, x = 'displ', y = 'hwy')
```

## Solución 1: jitter 
* usa `lmplot` si tus datos son continuos y `stripplot` si son categóricos 

```python
sns.lmplot(data = mpg, x = 'displ', y = 'hwy', 
fit_reg = False, x_jitter=0.1, y_jitter=0.1)
```

## Solución 2: `alpha`

```python
sns.scatterplot(data = mpg, x = 'displ', y = 'hwy', 
alpha = 0.6)
```

## ¿Están diciendo lo mismo estas dos gráficas?

```python
import statsmodels
fig, (ax1, ax2) = plt.subplots(nrows = 1, ncols = 2)
sns.regplot(data = mpg, x = 'displ', y = 'hwy', fit_reg = False, ax = ax1)
sns.regplot(data = mpg, x = 'displ', y = 'hwy', scatter = False, lowess = True,ci=None,  ax = ax2)
plt.show()
```

## Relación entre el tamaño del motor y las millas por galón

```python
fig, (ax1, ax2) = plt.subplots(nrows = 1, ncols = 2)
sns.regplot(data = mpg, x = 'displ', y = 'hwy', 
fit_reg = False, ax = ax1)
sns.regplot(data = mpg, x = 'displ', y = 'hwy', 
scatter = False, lowess = True,ci=None, 
ax = ax2)
plt.show()
```

## Podemos ver estas dos gráficas juntas usando `regplot` o `lmplot`

```python
sns.regplot(data = mpg, x = 'displ', y = 'hwy', 
lowess = True, scatter_kws={"color": "blue"}, 
line_kws={"color": "red"})
```

## Differencias entre `regplot` y `lmplot`

* `lmplot` combina `regplot` y facets: esta pensado para sacar regresiones en subconjuntos de los datos
* `regplot` solo puedes hacer la regresión en todos los datos, si quieres dividir tendrías que manipular los datos antes 


## `lmplot` puede hacer lo siguiente

```python
sns.lmplot(data = mpg, x = 'displ', y = 'hwy',
hue = 'drv', order = 2, scatter = False)
```

## ¿Qué estan haciendo `regplot` y `lmplot`?
* Agregar una linea de regresión 
* El método de default es una regresión lineal 
    * Polynomial regression:  `order = n` donde n son los grados de tu polinomio 
    * Lowess: locally weighted linear regresión no paramétrica `lowess = True` (necesitas tener la librería `statsmodels` instalada para que funcione)
    * Logaritmo: `logx = True` estima $y \sim log(x)$
    * Logistic: `logistic = True` y es binaria también necesitas `statsmodel`

# Transformaciones estadísticas

## ¿Qué está haciendo esta gráfica?

```python
char = sns.countplot(data = mpg, x = 'manufacturer')
char.set_xticklabels(labels = mpg.manufacturer.unique(), rotation = 45, ha = 'right')
```

## ¿Qué está haciendo esta gráfica?
* Contesta la pregunta: ¿Cuántos autos tenemos por marca?
* Pero, no tenemos la variable: número de autos por marca en la base
* Varias opciones para graficar esto 

## Opción 1: explícitamente calcular los valores 

```python
mpg.manufacturer.value_counts().plot.bar()
## Seaborn 
contar_marcas = mpg.manufacturer.value_counts().rename_axis('manufacturer').reset_index(name='counts')
sns.barplot(data = contar_marcas, x = 'manufacturer', y = 'counts')
plt.xticks(rotation = 45, ha = 'right')
plt.show()
```

## Opcion 2: usar `countplot` de `seaborn`
* La mayoría de los plots que usamos grafican _raw data _
* Algunos calculan variables nuevas que no están en nuestra base
    * `countplot`, histogramas, frequency polygons: ponen los datos en cajas y luego grafican lo que hay en cada caja  
    * smoothers: calculan un modelo 
    * boxplots: calculan resúmenes de la distribución

## `countplot`
* Calcula cuántas observaciones hay por grupo por nosotros 

```python
sns.countplot(data = mpg, x = 'manufacturer')
plt.xticks(rotation = 45, ha = 'right')
plt.show()
```

## `boxplot`

```python
sns.boxplot(data = mpg, x = "class", y = "hwy")
```

## Ordenando el boxplot 

```python
my_order = mpg.groupby('class')['hwy'].median().sort_values().index
sns.boxplot(data = mpg, x = 'class', y = 'hwy', order = my_order)
```

## Subplots: Figura

## Ejemplo:

```python
fig, [[ax1, ax2], [ax3, ax4]] = plt.subplots(nrows = 2, ncols = 2, constrained_layout=True)
fig.suptitle('Millas por galones por tamaño del motor y año de fabricación')
ax1.set_title('1999')
ax2.set_title('2008')
ax3.set_title('Proporción de autos por tipo')
ax4.set_title('Cantidad por marca')
mpg[mpg.year == 1999].plot.scatter(x = 'displ', y = 'hwy', c = 'red', alpha = 0.5, ax = ax1)
mpg[mpg.year == 2008].plot.scatter(x = 'displ', y = 'hwy', c = 'red', alpha = 0.5, ax = ax2)
mpg.groupby('class').size().plot.pie(ax = ax3)
mpg.manufacturer.value_counts().plot.bar(ax = ax4)
plt.show()
```

## ¡¡CUIDADO!!!

* Los subplots y los facets no son lo mismo 
    * Facet: dividir la gráfica por una variable (similar a un color/figura por cada grupo)
    * Subplot: 
        * general
        * no necesitas que los datos estén en la misma base
        * no necesitas que sea el mismo tipo de gráfico/ imagen
        * no necesitas que las divisiones sean iguales 

## ¡¡CUIDADO 2!!

* Usamos `import matplotlib.pyplot as plt` para importar el modulo que necesitamos de `matplotlib`
* Casi todas las funciones de `pyplot` se refieren a una figura o un eje (axes) que ya existe o crea uno nuevo
* ¿Qué quiere decir esto?
    * llamar a `plt.` se refiere a la última figura o eje que usaste
    * si no quieres modificar ese mejor declara un eje nuevo y modifica este 
    
## Ejemplo: muy fácil

### Ejemplo: ahora quiero una nueva gráfica 

```python
sns.countplot(data = mpg, x = 'manufacturer')
plt.xticks(rotation = 45, ha = 'right')
plt.show()
```

### ¿Qué paso acá?

* `plt.show()` está mostrando la primera gráfica que hicimos
* Para solucionarlo necesitamos declarar una nueva figura 
* OJO: en notebooks y en editores de python esto normalmente no va a pasar pero si usan `.py` van a necesitar declarar su figura para que esto no pase 
* Es buena idea siempre declarar una nueva figura!!

### Solución: crea otra figura 

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax = sns.countplot(data = mpg, x = 'manufacturer')
ax.set_xticklabels(labels = mpg.manufacturer.unique(), rotation = 45, ha = 'right')
plt.show()
```
