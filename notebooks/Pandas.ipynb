### 2. Indices

¿Qué operaciones quiero hacer con los datos? ¿Cómo los quiero escribir?
Los DataFrame en pandas tienen dos ejes(axis) de indices: columnas y renglones.
Para este ejemplo vamos a usar la base de datos de los aviones `planes.csv`

```python
#1. ¿Cómo se ve esta base?
planes = pd.read_csv(raw_dir/'planes.csv')
planes
```

```python
#2. Definir los indices. En este caso el número del avión
planes = planes.set_index(pd.CategoricalIndex(planes['plane'], name = 'identifier'))
## tiramos la columna original del numero de avión
planes = planes.drop('plane', axis = 'columns')
```

```python
#3. Explorar esta base 
print(planes.columns)
print(planes.index)
```

#### ¿Qué podemos hacer sobre los indices?
* `df.set_axis`: establece el eje, puede funcionar para cambiar el nombre de las columnas
* `df.rename_axis`: renombra el eje 
* `df.rename`: renombra la etiqueta de algún indice (podemos usarlo para cambiar el nombre de las columnas)
* `df.reindex`: reindexa los renglones (cambiando la frecuencia o agregando renglones) 

```python
## Ejemplos de lo anterior: 
dta = pd.DataFrame({"A":[1,2,3], "B":[4,5,6]})
print(dta)
print("Cambiamos los indices del defaul a: a,b,c")
print(dta.set_axis(['a','b','c'], axis = 'index'))
print("Cambiamos las columnas de A,B a C y D")
print(dta.set_axis(['C', 'D'], axis = 'columns'))
print("Le ponemos nombre al indice")
print(dta.rename_axis('new_index'))
print("Le cambiamos el nombre a numeros")
print(dta.rename_axis('numeros', axis = 'columns'))
print("Le cambiamos el nombre a las columnas")
print(dta.rename(columns = {'A':'columna_1', 'B':'columna_2'}))

index = ['a', 'b', 'c']
dta2 = pd.DataFrame({"A":[1,2,3], "B":[4,5,6]},index = index)
index_bis = ['a', 'b', 'c', 'd']
dta2.reindex(index_bis)
```

### 2. SELECCIONAR

Una de las primeras tareas durante el procesamiento y exploración de datos es la selección de datos. Pandas a través de los DataFrames nos proveen de herramientas para la selección de columnas y registros de manera eficiente.

* Operaciones para seleccionar renglones:
   * `df.loc`
   * `df.iloc`
   * `IndexSlide`
   * `df[mask]`
   * `df.query`
   * `df.sample`

#### 2.1 .loc

```python
# .loc trabaja con los indices
# nos regresa 3 observaciones de la base planes
planes.sample(3)
# Nos regresa el avion con el indice: N937DL
planes.loc['N937DL']
# Nos regresa dos aviones con los indices: 'N708SW' y'N253WN'
planes.loc[['N708SW', 'N253WN']]
# Nos regresa los primeros 3 aviones despues de acomodar los indices
planes.sort_index().head(3)
# Nos regresa todos los indices entre el 'N11106'y el'N11119'
planes.sort_index().loc['N11106':'N11119'] #¡Es cerrado!
# Nos regresa todos los aviones entre N11106 y N11119 pero solo las columnas entre model y speed
planes.sort_index().loc['N11106':'N11119', 'model':'speed']
```

#### 2.2 .iloc

```python
# .iloc trabaja con posiciones 
# nos regresa el 3 elemento de la base (el conteo empieza en 0)
planes.iloc[3]
# nos regresa el último elemento de la base planes
planes.iloc[-1]
# nos regresa el elemento 1, el 3 y el 7
planes.iloc[[1,3,7]]
# nos regresa entre el elemento cuatro y el siete!
planes.iloc[4:8]
# nos regresa entre el elemento cuatro y el seis con las columnas 2 y 0
planes.iloc[4:7, [2,0]]
#planes.iloc[2
```

#### 2.3 df[mask]

```python
# nos regresa solo la columna year
planes['year']
# nos regresa una serie booleana si se cumple o no la condicion year = 2000
planes['year'] == 2000
# nos regresa todas las observaciones con year = 2000 (158 observaciones)
planes[planes['year'] == 2000]
# nos regresa todas las observaciones con year = 2000 y mfr = AIRBUS INDUSTRIE
planes[(planes['year'] == 2000) & (planes['mfr'] == 'AIRBUS INDUSTRIE')]
# crea una condicion para filtrar los datos
old_planes_mask = planes['year'] < 1990
# filtra de acuerdo a la condición old_places_mask
planes[old_planes_mask]
# selecciona la columna year y filtra todos los aviones con año despues de 1990 
planes.loc[:,'year'].loc[lambda col: col > 1990]
```

#### 2.4 filter: filtra usando el indice

```python
# nos quedamos con todos los aviones que en su identifier tienen AA
planes.filter(like = 'AA', axis = 'index')
# nos quedamos con los dos aviones con identifier que conincidan con N492AA y N482AA
planes.filter(regex = 'N492AA|N482AA', axis = 'index')
```

#### 2.5 query

```python
engines = ['Turbo-shaft', '4 Cycle']
# nos regresa los aviones con engine turbo-shaft y velocidad > 108
planes.query('engine == "Turbo-shaft" and speed > 108')
# nos regresa los aviones con engines en la lista y velocidad > 108
planes.query('engine.isin(@engines) and speed >108')
```

### 4. Operaciones con una sola fuente de datos 
1. `.agg` agregar sobre alguno de los ejes 

```python
# para una sola variable
planes['speed'].agg(['mean','max'])
# para varias variables
planes.agg({'speed':['mean', 'max'], 'year':['max','min']})
```

2. `.transform` no cambia el tamaño del dataframe si empezamos con n renglones terminamos con n renglones

```python
# Por ejemplo: si queremos saber la edad de los aviones 
planes.year.transform(lambda x: 2023 - x)
```

3. `.apply` aplica una función a lo largo de un eje y transforma el dataframe a cualquier tamaño 

```python
def plane_size(number):
    if number > 80:
        return 'large'
    return 'small'

planes['no.seats'].apply(plane_size)
```

**Ojo:** apply y transform se parecen pero no son iguales

`transform`
   * puede recibir una funcion, una string function, una lista de funciones y un diccionario
   * no puede producir resultados agregados (si le das n renglones, regresa n renglones)
   * solo puede trabajar con una Serie 
   
`apply`
   * solo puede recibir funciones
   * puede producir resultados agregados (le das n renglones, puede regresar <n reglones)
   * puede trabajar con muchas series al mismo tiempo    

Estructura de estas operaciones 
1. `Dataframe.operacion(args)`
2. Siempre regresean una Serie o un DataFrame nuevo
3. Por default nunca modifica, si quieres que modifique necesitas:
     * `inplace = True`
     *  assignar tu operacion a una nueva columna o dataframe

#### 4.3 Agrupaciones

Algunas veces quieres calcular algo como `z = (x - mean(x))/sd(x)`

* Queremos encontrar que tan rápido van los aviones por tipo de motor. 
* ¿Qué necesitamos hacer para contestar esta pregunta?
    * Agrupar los aviones por tipo de motor 
    * Sacar la media de la velocidad por cada tipo de motor 
* ¿Cómo hacemos esto? `groupby()`

```python
# primero agrupamos por motor, le indicamos que variable nos interes (speed) y 
# sacamos lo que queremos en este caso la media
# Ojo: siempre nos va a regresar una serie
planes.groupby('engine')['speed'].mean()
```

* `groupby` proviene de un comando en el lenguaje de base de datos SQL
* Quizás sea más esclarecedor pensar en él en los términos acuñados por primera vez por Hadley Wickham: dividir (split), aplicar (apply), combinar (combine).
* Observa la siguiente imagen:
![GroupBy](https://jakevdp.github.io/PythonDataScienceHandbook/figures/03.08-split-apply-combine.png)

El agrupamiento ocurre como sigue:

* El paso de división implica dividir y agrupar un DataFrame según el valor de la clave especificada.
* El paso de aplicación implica calcular alguna función, generalmente un agregado, transformación o filtrado, dentro de los grupos individuales.
* El paso de combinación combina los resultados de estas operaciones en una matriz de salida.

#### 4.4 Agrupaciones con agregaciones

Por ejemplo si queremos encontrar cuanto se retrasan los vuelos por compañia

```python
flights.groupby('carrier')['dep_delay'].agg(['mean', 'min', 'max'])
```

 En el caso anterior, hemos indicado una lista con las agregaciones deseadas. Sin embargo, también es posible pasar un diccionario con el nombre de las columnas, y las operaciones a ser aplicadas.

```python
# ¿Qué estamos calculando acá?
flights.groupby('carrier').agg({'dep_delay':['mean','min','max'],
                               'dest':'nunique'})
```
